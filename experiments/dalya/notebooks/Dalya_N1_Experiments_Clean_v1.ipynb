```python
from google.colab import drive
import stat as _stat

drive.mount("/content/drive")

def _drive_io(operacao, descricao):
    """Retry transient I/O errors only, with at most one remount."""
    global _DRIVE_REMOUNT_USED
    transitorios = {
        errno.EIO, errno.ECONNABORTED, errno.ECONNRESET,
        errno.ENOTCONN, errno.ETIMEDOUT, getattr(errno, "ESTALE", 116),
    }
    for tentativa in range(_IO_RETRIES):
        try:
            return operacao()
        except OSError as exc:
            if exc.errno not in transitorios or tentativa == _IO_RETRIES - 1:
                raise OSError(
                    exc.errno,
                    f"{descricao}: {exc.strerror or str(exc)}. "
                    "Setup stopped. Existing data and results have been preserved.",
                    getattr(exc, "filename", None),
                ) from exc
            print(
                f"Temporary Drive error: {descricao}. "
                f"Retry {tentativa + 2}/{_IO_RETRIES}.",
                flush=True,
            )
            if exc.errno in {errno.ECONNABORTED, errno.ENOTCONN, errno.ECONNRESET}:
                if not _DRIVE_REMOUNT_USED:
                    _DRIVE_REMOUNT_USED = True
                    _DRIVER_RECOVERIES.append(descricao)
                    drive.mount("/content/drive", force_remount=True)
            time.sleep(1)

def _file_signature(caminho):
    caminho = Path(caminho)
    st = _drive_io(caminho.stat, f"check {caminho.name}")
    return {"name": caminho.name, "bytes": st.st_size, "mtime_ns": st.st_mtime_ns}

def _drive_is_file(caminho):
    caminho = Path(caminho)
    try:
        st = _drive_io(caminho.stat, f"check {caminho.name}")
    except FileNotFoundError:
        return False
    return _stat.S_ISREG(st.st_mode)

def _local_copy(caminho, grupo="files"):
    """Temporary Colab copy. The source on Drive is read-only."""
    caminho = Path(caminho)
    assinatura = _file_signature(caminho)
    chave = hashlib.sha256(
        json.dumps([str(caminho), assinatura], sort_keys=True).encode()
    ).hexdigest()[:20]
    destino = LOCAL_SESSION_DIR / grupo / f"{chave}_{caminho.name}"
    destino.parent.mkdir(parents=True, exist_ok=True)

    if destino.is_file() and destino.stat().st_size == assinatura["bytes"]:
        return destino

    def copiar():
        temporario = destino.with_suffix(destino.suffix + ".tmp")
        shutil.copyfile(caminho, temporario)
        assert temporario.stat().st_size == assinatura["bytes"], (
            f"Incomplete copy: {caminho.name}"
        )
        assert _file_signature(caminho) == assinatura, (
            f"The file changed while it was being read: {caminho.name}"
        )
        temporario.replace(destino)
        return destino

    return _drive_io(copiar, f"copy {caminho.name} to local disk")

def _read_json(caminho):
    local = _local_copy(caminho, "reports")
    return json.loads(local.read_text(encoding="utf-8"))

def _atomic_bytes(caminho, conteudo):
    caminho = Path(caminho)
    def escrever():
        caminho.parent.mkdir(parents=True, exist_ok=True)
        temporario = caminho.with_suffix(caminho.suffix + ".tmp")
        temporario.write_bytes(conteudo)
        temporario.replace(caminho)
    _drive_io(escrever, f"save {caminho.name}")

def converter_numpy(obj):
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    if isinstance(obj, np.generic):
        return obj.item()
    raise TypeError(f"Unsupported type: {type(obj).__name__}")

def _atomic_json(caminho, objeto):
    conteudo = json.dumps(
        objeto, default=converter_numpy, ensure_ascii=False, indent=2
    ).encode("utf-8")
    _atomic_bytes(caminho, conteudo)

assert _stat.S_ISDIR(_drive_io(Path(DATA).stat, "open the data folder").st_mode), DATA
assert _stat.S_ISDIR(_drive_io(TRAIN_DIR.stat, "open the training folder").st_mode), TRAIN_DIR
print("Drive is mounted and the data folder is accessible.")
_SETUP_STAGES_68.add("drive")
```

```python
_require_stage_68("features")
EPOCH_TIMES_VERSION_68 = 1
EPOCH_SECONDS_68 = 30.0
CONSECUTIVE_TOL_68 = 1e-6
RK_TEMPOS_68 = {
    "Sleep stage W": "W", "Sleep stage 1": "N1", "Sleep stage 2": "N2",
    "Sleep stage 3": "N3", "Sleep stage 4": "N3", "Sleep stage R": "REM",
}
META_TEMPOS_68 = {
    "version": EPOCH_TIMES_VERSION_68,
    "course_revision": COURSE_REVISION,
    "epoch_seconds": EPOCH_SECONDS_68,
    "required_native_fs": 100.0,
    "time_origin": "PSG sample zero",
    "event_rounding": True,
    "stage_mapping": RK_TEMPOS_68,
    "mne": mne.__version__,
    "numpy": np.__version__,
}
FICHEIRO_TEMPOS_68 = Path(CACHE_VIGGY) / "viggy_epoch_times_68_v1.json"
_LOCAL_TIMES_PENDING = LOCAL_SESSION_DIR / "epoch_times_pending_v1.json"

def _ler_tempos_edf_68(psg, hyp_local, labels):
    # Read the PSG header and use a small local copy of the hypnogram.
    raw = mne.io.read_raw_edf(psg, preload=False, verbose="ERROR")
    try:
        fs = float(raw.info["sfreq"])
        assert abs(fs - 100.0) < 1e-6 and raw.first_samp == 0
        raw.set_annotations(mne.read_annotations(hyp_local), emit_warning=False)
        eventos, ids = mne.events_from_annotations(
            raw,
            event_id={k: i for i, k in enumerate(RK_TEMPOS_68)},
            use_rounding=True, chunk_duration=EPOCH_SECONDS_68, verbose="ERROR",
        )
        assert np.all(eventos[:, 0] >= 0)
        passo = int(EPOCH_SECONDS_68 * fs)
        eventos = eventos[eventos[:, 0] + passo <= raw.n_times]
        inverso = {codigo: RK_TEMPOS_68[nome] for nome, codigo in ids.items()}
        lidas = np.asarray([inverso[int(e[2])] for e in eventos]).astype(str)
        np.testing.assert_array_equal(lidas, labels, err_msg=f"Labels: {psg.name}")
        return eventos[:, 0].astype(float) / fs, float(raw.n_times / fs)
    finally:
        raw.close()

def _validar_tempos_68(tempos, duracao, labels):
    assert tempos.shape == labels.shape and len(labels) > 0
    assert np.isfinite(tempos).all() and np.isfinite(duracao)
    assert np.all(np.diff(tempos) > 0) and tempos[0] >= 0
    assert tempos[-1] + EPOCH_SECONDS_68 <= duracao + CONSECUTIVE_TOL_68

def preparar_tempos_sessao_68():
    cache = {"metadata": META_TEMPOS_68, "records": {}}
    pendente_reposto = False
    if _drive_is_file(FICHEIRO_TEMPOS_68):
        anterior = _read_json(FICHEIRO_TEMPOS_68)
        if anterior.get("metadata") == META_TEMPOS_68:
            cache = anterior
    if _LOCAL_TIMES_PENDING.is_file():
        pendente = json.loads(_LOCAL_TIMES_PENDING.read_text(encoding="utf-8"))
        if pendente.get("metadata") == META_TEMPOS_68:
            cache["records"].update(pendente["records"])
            pendente_reposto = True

    tempos_por_gravacao, blocos_t, blocos_prev = {}, [], []
    calculadas = reutilizadas = posicao = 0
    alterado = pendente_reposto

    try:
        for numero, (psg, dados) in enumerate(zip(ficheiros_b0_68, b0_gravacoes_68), 1):
            record = identificar_gravacao_reposta_68(dados["record"])
            assert psg.name == f"{record}-PSG.edf"
            assert record[:5] in DEVELOPMENT_GROUPS_68
            labels = np.asarray(dados["y"]).astype(str)

            hyp = psg.with_name(psg.name.replace("-PSG.edf", "-Hypnogram.edf"))
            if not _drive_is_file(hyp):
                candidatos = _drive_io(
                    lambda: sorted(psg.parent.glob(f"{record[:6]}*Hypnogram.edf")),
                    f"locate annotations for {record}",
                )
                assert len(candidatos) == 1, f"Ambiguous annotations: {record}"
                hyp = candidatos[0]

            assinatura = {
                "psg": _file_signature(psg),
                "hypnogram": _file_signature(hyp),
                "labels_sha256": hashlib.sha256("\n".join(labels).encode()).hexdigest(),
            }
            entrada = cache["records"].get(record)
            if entrada is not None and entrada.get("signature") == assinatura:
                tempos = np.asarray(entrada["onsets_seconds"], dtype=float)
                duracao = float(entrada["duration_seconds"])
                reutilizadas += 1
            else:
                hyp_local = _local_copy(hyp, "annotations")
                tempos, duracao = _drive_io(
                    lambda: _ler_tempos_edf_68(psg, hyp_local, labels),
                    f"read the header for {record}",
                )
                _validar_tempos_68(tempos, duracao, labels)
                assert _file_signature(psg) == assinatura["psg"]
                assert _file_signature(hyp) == assinatura["hypnogram"]
                cache["records"][record] = {
                    "signature": assinatura,
                    "onsets_seconds": tempos.tolist(),
                    "duration_seconds": duracao,
                }
                calculadas += 1
                alterado = True

            _validar_tempos_68(tempos, duracao, labels)
            fim = posicao + len(labels)
            assert np.all(record_ids_b0_68[posicao:fim] == record)
            np.testing.assert_array_equal(y_b0_68[posicao:fim], labels)
            posicao = fim

            anterior_valida = np.zeros(len(tempos), dtype=bool)
            anterior_valida[1:] = np.isclose(
                np.diff(tempos), EPOCH_SECONDS_68, rtol=0, atol=CONSECUTIVE_TOL_68
            )
            tempos_por_gravacao[record] = (tempos, duracao)
            blocos_t.append(tempos)
            blocos_prev.append(anterior_valida)

            # Save the first recording, then save at regular checkpoints.
            if numero == 1 or numero % 25 == 0 or numero == len(ficheiros_b0_68):
                if alterado:
                    _atomic_json(FICHEIRO_TEMPOS_68, cache)
                    alterado = False
                if numero % 25 == 0 or numero == len(ficheiros_b0_68):
                    print(f"Epoch times checked: {numero}/{len(ficheiros_b0_68)}", flush=True)
    except (Exception, KeyboardInterrupt):
        # Keep the session work if Drive becomes unavailable.
        _LOCAL_TIMES_PENDING.write_text(
            json.dumps(cache, ensure_ascii=False), encoding="utf-8"
        )
        raise

    assert posicao == len(y_b0_68)
    if _LOCAL_TIMES_PENDING.is_file():
        _LOCAL_TIMES_PENDING.unlink()  # Only the temporary file on local disk.
    return tempos_por_gravacao, np.concatenate(blocos_t), np.concatenate(blocos_prev), calculadas, reutilizadas

(
    tempos_epocas_68, onsets_epocas_68, anterior_consecutiva_68,
    calculados_tempos_68, reutilizados_tempos_68,
) = preparar_tempos_sessao_68()

# Keep the legacy name for compatibility. This does not apply a Wake crop.
tempos_c1_68 = tempos_epocas_68
print(
    f"Epoch times ready: {len(tempos_epocas_68)} recordings | "
    f"{calculados_tempos_68} read now | {reutilizados_tempos_68} reused."
)
print("Epoch-time cache:", FICHEIRO_TEMPOS_68)
_SETUP_STAGES_68.add("times")
```

```python
# Dalya batch: executes only after SESSION READY. Existing team experiment files are read-only.
RUN_DALYA_BATCH = False
DALYA_BATCH_VERSION = "v1"
DALYA_OUT = Path("/content/drive/MyDrive/CM2013_Dalya_experiments") / DALYA_BATCH_VERSION

if RUN_DALYA_BATCH:
    import json as _json
    import csv as _csv
    from datetime import datetime as _datetime, timezone as _timezone
    from sklearn.base import clone as _clone
    from sklearn.metrics import (cohen_kappa_score as _kappa,
        f1_score as _f1, accuracy_score as _accuracy,
        precision_score as _precision, recall_score as _recall)

    assert SESSAO_PRONTA, "Run section 1 until SESSION READY before this cell."
    assert X_tc1_68.shape == (360621, 33)
    assert not set(groups_b0_68).intersection(HOLDOUT_GROUPS)
    assert FOLDS_68 == EXPECTED_FOLDS_68
    DALYA_OUT.mkdir(parents=True, exist_ok=True)
    _stages = ["W", "N1", "N2", "N3", "REM"]
    _y = np.asarray(y_b0_68, dtype=str)
    _rid = np.asarray(record_ids_b0_68, dtype=str)
    _baseline_name = "TC1_C1_68_crop30_fixed_AE"
    _baseline_pred = ler_previsoes_guardadas_68(_baseline_name)
    assert _baseline_pred is not None, "Missing saved TC1+C1 predictions in the shared experiments folder."
    _baseline_pred = np.asarray(_baseline_pred, dtype=str)
    assert len(_baseline_pred) == len(_y)
    _crop = {}
    for _fold in "ABCDE":
        _c = _read_json(EXPERIMENTS_DIR / f"C1_68_crop30_fixed_AE_fold_{_fold}.json")
        _tr = np.asarray(_c["train_indices"], dtype=int)
        _orig, _te = SPLITS_68[_fold]
        np.testing.assert_array_equal(np.asarray(_c["test_indices"], dtype=int), _te)
        assert np.isin(_tr, _orig).all() and np.intersect1d(_tr, _te).size == 0
        assert np.all(_y[np.setdiff1d(_orig, _tr)] == "W")
        assert set(groups_b0_68[_tr]) == set(groups_b0_68[_orig])
        assert set(groups_b0_68[_te]) == set(FOLDS_68[_fold])
        _crop[_fold] = (_tr, _te)

    def _scores(_p):
        _p = np.asarray(_p, dtype=str)
        assert len(_p) == len(_y) and np.isin(_p, _stages).all()
        _n2 = _stages.index("N2")
        return {
            "kappa": float(_kappa(_y, _p, labels=_stages)),
            "macro_f1": float(_f1(_y, _p, labels=_stages, average="macro", zero_division=0)),
            "accuracy": float(_accuracy(_y, _p)),
            "n1_recall": float(_recall(_y, _p, labels=["N1"], average="macro", zero_division=0)),
            "n1_precision": float(_precision(_y, _p, labels=["N1"], average="macro", zero_division=0)),
            "n1_f1": float(_f1(_y, _p, labels=["N1"], average="macro", zero_division=0)),
            "n2_f1": float(_f1(_y, _p, labels=["N2"], average="macro", zero_division=0)),
        }

    _rows = []
    def _record(_name, _pred, _status="completed", _notes=""):
        _m = _scores(_pred) if _pred is not None else {}
        _row = {"experiment": _name, "status": _status, **_m, "notes": _notes}
        _rows.append(_row)
        _path = DALYA_OUT / "comparison.csv"
        _cols = ["experiment", "status", "kappa", "macro_f1", "accuracy",
                 "n1_recall", "n1_precision", "n1_f1", "n2_f1", "notes"]
        with open(_path, "w", newline="", encoding="utf-8") as _f:
            _writer = _csv.DictWriter(_f, fieldnames=_cols)
            _writer.writeheader(); _writer.writerows(_rows)
        (DALYA_OUT / "comparison.json").write_text(
            _json.dumps(_rows, ensure_ascii=False, indent=2), encoding="utf-8")
        print(_name, _status, {k: round(v, 4) for k, v in _m.items()}, flush=True)

    def _run_rf(_tag, _x, _model):
        _x = np.asarray(_x)
        assert _x.shape[0] == len(_y) and np.isfinite(_x).all()
        _pred = np.empty_like(_y)
        _seen = np.zeros(len(_y), dtype=bool)
        for _fold in "ABCDE":
            _tr, _te = _crop[_fold]
            _save = DALYA_OUT / f"{_tag}_fold_{_fold}.npz"
            if _save.exists():
                with np.load(_save, allow_pickle=False) as _cached:
                    np.testing.assert_array_equal(_cached["test_indices"], _te)
                    _fold_pred = _cached["pred"].astype(str)
                print(_tag, _fold, "reused checkpoint", flush=True)
            else:
                print(_tag, _fold, "training...", flush=True)
                _clf = _clone(_model)
                _clf.fit(_x[_tr], _y[_tr])
                _fold_pred = np.asarray(_clf.predict(_x[_te]), dtype=str)
                _tmp = DALYA_OUT / f"{_tag}_fold_{_fold}.partial.npz"
                np.savez_compressed(_tmp, test_indices=_te, pred=_fold_pred)
                _tmp.replace(_save)
                print(_tag, _fold, "saved", flush=True)
            assert len(_fold_pred) == len(_te)
            assert np.isin(_fold_pred, _stages).all()
            assert not _seen[_te].any()
            _pred[_te] = _fold_pred
            _seen[_te] = True
        assert _seen.all()
        return _pred

    _record("TC1_C1_reference", _baseline_pred, _notes="Vicente's saved out-of-fold predictions; no retraining")

    # 1. Prespecified isolated N2 gap bridging between consecutive predicted N1 epochs.
    _transition = _baseline_pred.copy()
    _adj = np.asarray(anterior_consecutiva_68, dtype=bool)
    _middle = np.arange(1, len(_y) - 1)
    _ok = (_adj[_middle] & _adj[_middle + 1] &
           (_rid[_middle - 1] == _rid[_middle]) & (_rid[_middle] == _rid[_middle + 1]))
    _idx = _middle[_ok & (_baseline_pred[_middle - 1] == "N1") &
        (_baseline_pred[_middle] == "N2") & (_baseline_pred[_middle + 1] == "N1")]
    _transition[_idx] = "N1"
    _record("01_transition_N1_bridge", _transition, _notes=f"Prespecified N1-N2-N1 bridge; changed {len(_idx)} epochs; no labels used to edit predictions")

    # 2. Relative EEG band power temporal differences. Existing F2 last five columns are relative bands.
    _relative = np.asarray(X_tc1_68[:, 11:16] - X_tc1_68[:, 27:32], dtype=float)
    _relative[~_adj] = 0.0
    _x_rel = np.column_stack([X_tc1_68, _relative])
    _pred_rel = _run_rf("02_relative_temporal_delta", _x_rel, modelo_base_f2_68)
    _record("02_relative_temporal_delta", _pred_rel, _notes="5 current-minus-previous relative EEG power features; cropped training; 38 columns")

    # 3. No validated spindle/K-complex detector or cached event features supplied.
    _record("03_spindles_kcomplex", None, "not_run",
            "Requires raw EEG, validated event detector and new cached event features; no fabricated results")

    # 4. Keep baseline balanced weighting, multiply N1 weight by 1.5 (a prespecified hypothesis).
    _weighted = _clone(modelo_base_f2_68)
    _counts = {c: int((_y[_crop["A"][0]] == c).sum()) for c in _stages}
    assert all(v > 0 for v in _counts.values())
    # Per-fold weights are computed on each fold's cropped TRAIN labels only.
    # Pass sample weights through Pipeline to avoid any test-label involvement.
    def _run_n1_weighted():
        _pred = np.empty_like(_y)
        _seen = np.zeros(len(_y), dtype=bool)
        for _fold in "ABCDE":
            _tr, _te = _crop[_fold]
            _save = DALYA_OUT / f"04_n1_weight15_fold_{_fold}.npz"
            if _save.exists():
                with np.load(_save, allow_pickle=False) as _z:
                    np.testing.assert_array_equal(_z["test_indices"], _te)
                    _p = _z["pred"].astype(str)
                print("04_n1_weight15", _fold, "reused checkpoint", flush=True)
            else:
                _model = _clone(_weighted)
                # Baseline RF already has class_weight='balanced'; sample_weight multiplies N1 by 1.5.
                _sw = np.where(_y[_tr] == "N1", 1.5, 1.0)
                print("04_n1_weight15", _fold, "training...", flush=True)
                _model.fit(X_tc1_68[_tr], _y[_tr], clf__sample_weight=_sw)
                _p = np.asarray(_model.predict(X_tc1_68[_te]), dtype=str)
                _tmp = DALYA_OUT / f"04_n1_weight15_fold_{_fold}.partial.npz"
                np.savez_compressed(_tmp, test_indices=_te, pred=_p)
                _tmp.replace(_save)
                print("04_n1_weight15", _fold, "saved", flush=True)
            assert not _seen[_te].any() and len(_p) == len(_te)
            _pred[_te] = _p
            _seen[_te] = True
        assert _seen.all()
        return _pred
    _pred_weight = _run_n1_weighted()
    _record("04_n1_weight15", _pred_weight, _notes="Balanced RF + 1.5x N1 sample weight; same 33 features and cropped train folds")

    # 5. Exploratory feature-extremeness audit; NOT an EEG artifact or signal-quality measurement.
    # Thresholds are derived from training feature distributions for each fold only.
    _audit_rows = []
    for _fold in "ABCDE":
        _tr, _te = _crop[_fold]
        _med = np.median(X_tc1_68[_tr, :16], axis=0)
        _q25, _q75 = np.percentile(X_tc1_68[_tr, :16], [25, 75], axis=0)
        _scale = np.maximum(_q75 - _q25, 1e-8)
        _ext = np.mean(np.abs((X_tc1_68[_te, :16] - _med) / _scale) > 3, axis=1)
        for _name, _mask in [("no_extreme_features", _ext == 0), ("one_or_more_extreme_features", _ext > 0)]:
            if not np.any(_mask):
                continue
            _a = _te[_mask]
            _audit_rows.append({"fold": _fold, "group": _name, "n_epochs": int(len(_a)),
                "kappa": float(_kappa(_y[_a], _baseline_pred[_a], labels=_stages)),
                "n1_recall": float(_recall(_y[_a], _baseline_pred[_a], labels=["N1"], average="macro", zero_division=0))})
    _audit = DALYA_OUT / "05_feature_extremeness_audit.csv"
    with open(_audit, "w", newline="", encoding="utf-8") as _f:
        _w = _csv.DictWriter(_f, fieldnames=["fold", "group", "n_epochs", "kappa", "n1_recall"])
        _w.writeheader(); _w.writerows(_audit_rows)
    _record("05_feature_extremeness_audit", None, "analysis_only",
            "Exploratory proxy, NOT true signal quality; see 05_feature_extremeness_audit.csv")

    _log = ["# Dalya — Sleep-EDF sequential experiments", "", "Baseline: TC1_C1_68_crop30_fixed_AE", "", "All test epochs and subject folds A-E retained; holdout untouched.", "", "| Experiment | Status | Kappa | Macro-F1 | N1 recall | N1 precision | N1 F1 | N2 F1 |", "|---|---|---:|---:|---:|---:|---:|---:|"]
    for _r in _rows:
        _fmt = lambda k: f"{_r[k]:.4f}" if k in _r else "—"
        _log.append(f"| {_r['experiment']} | {_r['status']} | {_fmt('kappa')} | {_fmt('macro_f1')} | {_fmt('n1_recall')} | {_fmt('n1_precision')} | {_fmt('n1_f1')} | {_fmt('n2_f1')} |")
    _log += ["", "Spindle/K-complex detection requires validated signal-level feature extraction.", "Feature-extremeness analysis is NOT artifact QC.", "No automatic model selection, no holdout testing, and no GitHub/ClickUp/Excel sync."]
    (DALYA_OUT / "RESULTS_DALYA.md").write_text("\n".join(_log) + "\n", encoding="utf-8")
    print("BATCH FINISHED. Results saved in", DALYA_OUT, flush=True)
    display(pd.DataFrame(_rows))
```

## 3. Rebuild documentation from saved predictions — no retraining
This step loads the baseline JSON from the **shared Drive experiment directory** rather than a temporary manually uploaded `/content` file. It uses the adjacency and recording arrays created in setup to reconstruct the N1 Bridge. Existing CSVs/PNGs in the documentation folder will be regenerated. It will fail clearly if the required checkpoints or original reports are missing.

```python
RUN_DOCUMENTATION = False  # Set True to regenerate CSVs and figures; no model fitting.
print("Documentation enabled:", RUN_DOCUMENTATION)
```

```python
if RUN_DOCUMENTATION:

    # ============================================================
    # DALYA v1 — COMPLETE EXPERIMENT DOCUMENTATION
    # No retraining | Horizontal colorbar | Save to Google Drive
    # ============================================================

    import json
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt

    from pathlib import Path
    from IPython.display import display
    from sklearn.metrics import (
        accuracy_score,
        cohen_kappa_score,
        f1_score,
        balanced_accuracy_score,
        precision_recall_fscore_support,
        confusion_matrix
    )

    # ------------------------------------------------------------
    # 1. Paths
    # ------------------------------------------------------------

    ROOT = Path(
        "/content/drive/MyDrive/CM2013_Dalya_experiments/v1"
    )

    BASE = EXPERIMENTS_DIR / "TC1_C1_68_crop30_fixed_AE.json"

    OUT = ROOT / "documentation"
    assert globals().get("SESSAO_PRONTA", False), "Run Setup cells first."
    OUT.mkdir(parents=True, exist_ok=True)

    assert ROOT.is_dir(), f"Missing experiment folder: {ROOT}"
    assert BASE.is_file(), f"Missing baseline JSON: {BASE}"

    # ------------------------------------------------------------
    # 2. Load and validate original baseline
    # ------------------------------------------------------------

    with BASE.open(encoding="utf-8") as f:
        source = json.load(f)

    report = source["report"]

    stages = ["W", "N1", "N2", "N3", "REM"]

    y = np.asarray(report["y_true"]).astype(str)
    baseline_pred = np.asarray(report["y_pred"]).astype(str)
    fold_labels = np.asarray(report["y_fold"]).astype(str)

    assert len(y) == len(baseline_pred) == len(fold_labels) == 360621
    assert set(y).issubset(stages)
    assert set(baseline_pred).issubset(stages)

    unique_folds = sorted(set(fold_labels))

    if set(unique_folds) == set("ABCDE"):
        fold_map = {f: f for f in "ABCDE"}

    elif set(unique_folds) in (
        set("01234"),
        set("12345")
    ):
        fold_map = dict(zip("ABCDE", unique_folds))

    else:
        raise ValueError(
            f"Unexpected fold labels: {unique_folds}"
        )

    cm_base = confusion_matrix(
        y, baseline_pred, labels=stages
    )

    np.testing.assert_array_equal(
        cm_base,
        np.asarray(report["confusion"])
    )

    # ------------------------------------------------------------
    # 3. Load saved experiment predictions
    # ------------------------------------------------------------

    def read_saved_predictions(prefix):

        pred = np.empty(len(y), dtype=object)
        seen = np.zeros(len(y), dtype=bool)

        for fold in "ABCDE":

            path = ROOT / f"{prefix}_fold_{fold}.npz"
            assert path.is_file(), f"Missing: {path}"

            with np.load(path, allow_pickle=False) as z:
                idx = np.asarray(
                    z["test_indices"], dtype=int
                )
                p = np.asarray(z["pred"]).astype(str)

            expected = np.flatnonzero(
                fold_labels == fold_map[fold]
            )

            np.testing.assert_array_equal(
                np.sort(idx),
                expected,
                err_msg=f"Fold {fold} mismatch"
            )

            assert len(idx) == len(p)
            assert not seen[idx].any()
            assert set(p).issubset(stages)

            pred[idx] = p
            seen[idx] = True

        assert seen.all()

        return pred.astype(str)


    predictions = {
        "TC1+C1 Baseline": baseline_pred,

        "Relative Temporal Delta":
            read_saved_predictions(
                "02_relative_temporal_delta"
            ),

        "N1 Weighting x1.5":
            read_saved_predictions(
                "04_n1_weight15"
            )
    }

    # ------------------------------------------------------------
    # 4. Recover N1 Transition Bridge, if possible
    # ------------------------------------------------------------

    bridge_available = all(
        name in globals()
        for name in [
            "anterior_consecutiva_68",
            "record_ids_b0_68"
        ]
    )
    assert bridge_available, "Run all Setup cells first; N1 Bridge requires adjacency and recording IDs."

    if bridge_available:

        adj = np.asarray(
            anterior_consecutiva_68, dtype=bool
        )

        rid = np.asarray(
            record_ids_b0_68
        ).astype(str)

        assert len(adj) == len(rid) == len(y)

        middle = np.arange(1, len(y) - 1)

        valid = (
            adj[middle]
            & adj[middle + 1]
            & (rid[middle - 1] == rid[middle])
            & (rid[middle] == rid[middle + 1])
        )

        changed = middle[
            valid
            & (baseline_pred[middle - 1] == "N1")
            & (baseline_pred[middle] == "N2")
            & (baseline_pred[middle + 1] == "N1")
        ]

        assert len(changed) == 213, (
            f"Expected 213 changed epochs; got {len(changed)}"
        )

        bridge_pred = baseline_pred.copy()
        bridge_pred[changed] = "N1"

        predictions["N1 Transition Bridge"] = bridge_pred

    else:
        print(
            "N1 Bridge: original adjacency arrays unavailable."
        )
        print(
            "Saved aggregate results will be used."
        )

    # ------------------------------------------------------------
    # 5. Compute complete performance metrics
    # ------------------------------------------------------------

    def get_metrics(true, pred):

        precision, recall, f1, _ = (
            precision_recall_fscore_support(
                true,
                pred,
                labels=stages,
                zero_division=0
            )
        )

        result = {
            "Accuracy": accuracy_score(true, pred),

            "Cohen kappa": cohen_kappa_score(
                true, pred, labels=stages
            ),

            "Macro-F1": f1_score(
                true,
                pred,
                labels=stages,
                average="macro",
                zero_division=0
            ),

            "Balanced accuracy":
                balanced_accuracy_score(true, pred)
        }

        for i, stage in enumerate(stages):

            result[f"{stage} Precision"] = precision[i]
            result[f"{stage} Recall"] = recall[i]
            result[f"{stage} F1"] = f1[i]

        return result


    all_metrics = {}
    fold_rows = []

    for name, pred in predictions.items():

        all_metrics[name] = get_metrics(y, pred)

        for fold in "ABCDE":

            idx = fold_labels == fold_map[fold]

            m = get_metrics(y[idx], pred[idx])

            fold_rows.append({
                "Experiment": name,
                "Fold": fold,
                "Epochs": int(idx.sum()),
                **m
            })

    # ------------------------------------------------------------
    # 6. Recover saved Bridge aggregate results if necessary
    # ------------------------------------------------------------

    comparison = pd.read_csv(
        ROOT / "comparison.csv"
    )

    if not bridge_available:

        row = comparison.loc[
            comparison["experiment"]
            == "01_transition_N1_bridge"
        ]

        assert len(row) == 1
        r = row.iloc[0]

        all_metrics["N1 Transition Bridge"] = {
            "Accuracy": float(r["accuracy"]),
            "Cohen kappa": float(r["kappa"]),
            "Macro-F1": float(r["macro_f1"]),
            "N1 Precision": float(r["n1_precision"]),
            "N1 Recall": float(r["n1_recall"]),
            "N1 F1": float(r["n1_f1"]),
            "N2 F1": float(r["n2_f1"])
        }

    # ------------------------------------------------------------
    # 7. Fold statistics and summary
    # ------------------------------------------------------------

    fold_df = pd.DataFrame(fold_rows)

    summary_rows = []

    for name, m in all_metrics.items():

        sub = fold_df[
            fold_df["Experiment"] == name
        ]

        summary_rows.append({
            "Experiment": name,
            **m,

            "Mean fold kappa":
                sub["Cohen kappa"].mean()
                if len(sub) == 5 else np.nan,

            "SD fold kappa":
                sub["Cohen kappa"].std(ddof=1)
                if len(sub) == 5 else np.nan,

            "Training (min)":
                "N/A - post-processing"
                if name == "N1 Transition Bridge"
                else "Not recorded"
        })

    summary = pd.DataFrame(
        summary_rows
    ).set_index("Experiment")

    # ------------------------------------------------------------
    # 8. Validate against original saved metrics
    # ------------------------------------------------------------

    names = {
        "TC1+C1 Baseline": "TC1_C1_reference",
        "N1 Transition Bridge": "01_transition_N1_bridge",
        "Relative Temporal Delta": "02_relative_temporal_delta",
        "N1 Weighting x1.5": "04_n1_weight15"
    }

    for name, original_name in names.items():

        old = comparison.loc[
            comparison["experiment"] == original_name
        ]

        assert len(old) == 1
        old = old.iloc[0]

        for new_key, old_key in [
            ("Accuracy", "accuracy"),
            ("Cohen kappa", "kappa"),
            ("Macro-F1", "macro_f1")
        ]:

            np.testing.assert_allclose(
                summary.loc[name, new_key],
                float(old[old_key]),
                atol=1e-6,
                rtol=0
            )

    # ------------------------------------------------------------
    # 9. Save metrics to Google Drive
    # ------------------------------------------------------------

    summary.to_csv(
        OUT / "v1_all_metrics.csv"
    )

    fold_df.to_csv(
        OUT / "v1_fold_metrics.csv",
        index=False
    )

    print("\n=== ALL EXPERIMENTS ===")
    display(summary.round(6))

    print("\n=== FOLD-BY-FOLD RESULTS ===")
    display(
        fold_df[
            [
                "Experiment",
                "Fold",
                "Cohen kappa",
                "Macro-F1",
                "Balanced accuracy",
                "N1 Recall",
                "N1 F1"
            ]
        ].round(6)
    )

    # ------------------------------------------------------------
    # 10. Confusion matrices — HORIZONTAL COLORBAR
    # ------------------------------------------------------------

    for name, pred in predictions.items():

        if name == "TC1+C1 Baseline":
            continue

        cm = confusion_matrix(
            y, pred, labels=stages
        )

        filename = name.replace(" ", "_")

        pd.DataFrame(
            cm,
            index=stages,
            columns=stages
        ).to_csv(
            OUT / f"{filename}_confusion_counts.csv"
        )

        # Two matrices with equal dimensions
        fig, axes = plt.subplots(
            1, 2,
            figsize=(12, 5.8),
            constrained_layout=True
        )

        for ax, matrix, title in [
            (axes[0], cm_base, "TC1+C1 Baseline"),
            (axes[1], cm, name)
        ]:

            normalized = (
                matrix
                / matrix.sum(axis=1, keepdims=True)
                * 100
            )

            im = ax.imshow(
                normalized,
                cmap="Blues",
                vmin=0,
                vmax=100,
                aspect="equal"
            )

            ax.set_xticks(range(5))
            ax.set_xticklabels(stages)

            ax.set_yticks(range(5))
            ax.set_yticklabels(stages)

            ax.set_xlabel("Predicted stage")
            ax.set_ylabel("True stage")
            ax.set_title(title, fontsize=12, pad=12)

            for i in range(5):
                for j in range(5):

                    value = normalized[i, j]

                    ax.text(
                        j, i,
                        f"{value:.1f}",
                        ha="center",
                        va="center",
                        fontsize=9,
                        color=(
                            "white" if value > 50
                            else "black"
                        )
                    )

        # One shared horizontal colorbar below both matrices
        cbar = fig.colorbar(
            im,
            ax=axes,
            orientation="horizontal",
            fraction=0.055,
            pad=0.12,
            shrink=0.65
        )

        cbar.set_label(
            "% of true sleep stage",
            fontsize=10
        )

        cbar.set_ticks(
            [0, 20, 40, 60, 80, 100]
        )

        fig.suptitle(
            f"Baseline vs {name}",
            fontsize=14,
            fontweight="bold"
        )

        save_path = (
            OUT / f"{filename}_comparison.png"
        )

        fig.savefig(
            save_path,
            dpi=200,
            bbox_inches="tight"
        )

        plt.show()
        plt.close(fig)

    # ------------------------------------------------------------
    # 11. Final confirmation
    # ------------------------------------------------------------

    print("\n===================================")
    print("DOCUMENTATION COMPLETED")
    print("===================================")
    print("Saved to Google Drive:")
    print(OUT)
    print("\nNo models were retrained.")
    print("Horizontal colorbars applied.")

else:
    print("Documentation skipped. Set RUN_DOCUMENTATION = True when ready.")
```
